# E4-C — Arsitektur, modif C: lebih dalam & lebih lebar + Dropout
**Interpretasi dari E4-B:** padding+BatchNorm menaikkan test acc 43.7% → **56.7%**, dan kurva train/val masih turun-turunnya di epoch 3 (train acc 52.1% vs val 57.0% — val lebih tinggi karena train diaugmentasi). Artinya: masih *underfitting*, bottleneck sekarang di **kapasitas** — hanya 3 conv sempit (16/32/64 channel).

**Modif C:**
- 4 conv block (32/64/128/128 channel) — representasi lebih kaya
- Head fc 128→256→10 (lebih besar dari 64→128→64→10)
- `Dropout(0.3)` sebelum fc — antisipasi overfitting karena kapasitas naik ~8x

Hyperparameter tetap SGD lr=0.0025, batch 10.
