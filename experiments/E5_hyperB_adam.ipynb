# E5-B — Hyperparameter, modif B: SGD → Adam
**Interpretasi dari E5-A:** lr 0.01 + batch 64 memang 4x lebih cepat (52s vs 211s), tapi test acc turun 66.6% → **61.8%**. Penyebab: total update turun dari 12.000 step (4000×3) jadi 1.875 step (625×3) — model undertrained; dan train acc (65.5%) kini > val acc (62.4%) → mulai sedikit overfit. SGD+momentum butuh banyak step untuk konvergen; dengan step sedikit, optimizer adaptif seharusnya lebih efisien.

**Modif B:** ganti optimizer **SGD → Adam** (lr=0.001, default β). Adam menyesuaikan learning rate per-parameter dari momentum orde-1 & 2, jadi tiap step lebih efektif. Batch tetap 64.
